```python
#movie id
#title
#overview
#genres
#keywords
#cast
#crew
```

```python
movies = movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
```

```python
movies.head()
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.isnull().sum()
```

```python
import ast

def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
#fetch the director from the crew column 
#get only first 3 costs present in the cast column
count = 0
for i in movies['crew']:
    count += 1
    print(i)
    if count <= 3:
        break
```
